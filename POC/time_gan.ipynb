The `timegan.py` file contains several TensorFlow 1.x constructs that are deprecated or unavailable in TensorFlow 2.x/Keras 3. Specifically, the following need to be addressed:

*   `tf.nn.rnn_cell.MultiRNNCell`: This is the direct cause of the `AttributeError`. It needs to be replaced with a Keras equivalent like `tf.keras.layers.StackedRNNCells` or by explicitly stacking multiple `tf.keras.layers.RNN` layers with individual `tf.keras.layers.GRUCell` or `tf.keras.layers.LSTMCell` instances.

*   `tf.contrib.layers.fully_connected`: This should be replaced with `tf.keras.layers.Dense`.
*   `tf.variable_scope`: This is a TensorFlow 1.x construct for managing variable reuse. In TensorFlow 2.x, variable reuse is typically handled by `tf.keras.layers.Layer` or `tf.keras.Model` objects, which manage their own variables, or by using `tf.Variable` directly within `tf.function`.

## Migration Strategy for `timegan.py` to TensorFlow 2.x/Keras

To address the `AttributeError: MultiRNNCell` and other TensorFlow 1.x deprecations, the `timegan.py` file will undergo a significant migration. The core strategy involves restructuring the model components as `tf.keras.Model` or `tf.keras.layers.Layer` subclasses, replacing TensorFlow 1.x specific operations with their TensorFlow 2.x Keras counterparts, and adapting the training loop for eager execution.

Here's a breakdown of the planned changes:

1.  **Model Architecture (Embedder, Recovery, Generator, Supervisor, Discriminator)**:
    *   Each of these components will be refactored into separate `tf.keras.Model` subclasses.
    *   `tf.nn.rnn_cell.MultiRNNCell` and `rnn_cell(module_name, hidden_dim)` will be replaced by stacked `tf.keras.layers.GRU` or `tf.keras.layers.LSTM` layers (depending on `module_name`). Keras layers inherently handle the stacking and dynamic unrolling.
    *   `tf.nn.dynamic_rnn` will be replaced by directly calling the Keras RNN layers, which handle the sequence processing.
    *   `tf.contrib.layers.fully_connected` will be replaced by `tf.keras.layers.Dense`.
    *   Input handling will switch from `tf.placeholder` to `tf.keras.Input` within the Keras models or by accepting direct tensor inputs in `call` methods.
    *   `tf.variable_scope` will be implicitly managed by the Keras `Model` or `Layer` classes, as they handle variable creation and reuse automatically.

2.  **Loss Functions**: `tf.losses.sigmoid_cross_entropy` and `tf.losses.mean_squared_error` will be replaced with their Keras equivalents: `tf.keras.losses.BinaryCrossentropy(from_logits=True)` and `tf.keras.losses.MeanSquaredError`.

3.  **Optimizers**: `tf.train.AdamOptimizer().minimize` will be replaced by `tf.keras.optimizers.Adam()`. The `minimize` call will be integrated into a custom training loop using `tf.GradientTape` for gradient computation and `optimizer.apply_gradients()` for weight updates.

4.  **Training Loop**: The `tf.Session()` and `sess.run()` based training loop will be completely overhauled. A custom training loop will be implemented using `tf.GradientTape` for automatic differentiation to compute gradients for all model components (Embedder, Generator, Discriminator, Supervisor).

5.  **Metrics**: `tf.nn.moments` will be replaced with standard TensorFlow 2.x tensor operations like `tf.reduce_mean` and `tf.math.reduce_variance`.

6.  **Initialization**: `tf.reset_default_graph()` will be removed as it is not relevant in TensorFlow 2.x eager execution context. Model weights will be initialized automatically by Keras or explicitly by calling the models with dummy input.

This refactoring will make the code compatible with current TensorFlow versions and leverage Keras's high-level API for model definition and training.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np

# Helper functions - adapted for TF2/Keras
def extract_time(data):
    """Returns max_seq_len and a list of sequence lengths for each sample."""
    time = list()
    max_seq_len = 0
    for i in range(len(data)):
        max_seq_len = max(max_seq_len, len(data[i]))
        time.append(len(data[i]))
    return time, max_seq_len

def random_generator(batch_size, z_dim, T_mb, max_seq_len):
    """Random vector generation.

    Args:
      - batch_size: size of the random vector
      - z_dim: dimension of the random vector
      - T_mb: time information (list of actual sequence lengths)
      - max_seq_len: maximum sequence length (for padding)

    Returns:
      - Z_mb: generated random vector (padded)
    """
    Z_mb = []
    for i in range(batch_size):
        temp = np.zeros([max_seq_len, z_dim], dtype=np.float32)
        temp_Z = np.random.uniform(0., 1, [T_mb[i], z_dim]).astype(np.float32)
        temp[:T_mb[i],:] = temp_Z
        Z_mb.append(temp)
    return tf.convert_to_tensor(Z_mb)

def batch_generator(data, time, batch_size):
    """Mini-batch generator.

    Args:
      - data: original time-series data (normalized and padded)
      - time: list of original sequence lengths
      - batch_size: mini-batch size

    Returns:
      - X_mb: time-series data in mini-batch (padded tensor)
      - T_mb: time information in mini-batch (list of original sequence lengths)
    """
    no = len(data)
    idx = np.random.permutation(no)
    train_idx = idx[:batch_size]

    X_mb_list = []
    T_mb_list = []
    for i in range(batch_size): # Fixed batch_idx to batch_size
        X_mb_list.append(data[train_idx[i]])
        T_mb_list.append(time[train_idx[i]])

    return tf.convert_to_tensor(X_mb_list), T_mb_list

def MinMaxScaler(data):
    """Min-Max Normalizer.

    Args:
      - data: raw data

    Returns:
      - norm_data: normalized data
      - min_val: minimum values (for renormalization)
      - max_val: maximum values (for renormalization)
    """
    min_val = np.min(np.min(data, axis = 0), axis = 0)
    data = data - min_val

    max_val = np.max(np.max(data, axis = 0), axis = 0)
    norm_data = data / (max_val + 1e-7)

    return norm_data, min_val, max_val

class TimeGANModel(keras.Model):
    def __init__(self, hidden_dim, num_layers, dim, module_name='gru', max_seq_len=None, gamma=1.0):
        super(TimeGANModel, self).__init__()
        self.hidden_dim = hidden_dim
        self.num_layers = num_layers
        self.dim = dim
        self.module_name = module_name
        self.max_seq_len = max_seq_len
        self.gamma = gamma

        # Models
        self.embedder_model = self._build_embedder()
        self.recovery_model = self._build_recovery()
        self.generator_model = self._build_generator()
        self.supervisor_model = self._build_supervisor()
        self.discriminator_model = self._build_discriminator()

        # Optimizers (will be instantiated later in the training loop)
        self.E0_optimizer = None
        self.E_optimizer = None
        self.G_optimizer = None
        self.GS_optimizer = None
        self.D_optimizer = None

        # Losses
        self.mse_loss = keras.losses.MeanSquaredError()
        self.bce_loss = keras.losses.BinaryCrossentropy(from_logits=True)

    def _build_embedder(self):
        inputs = keras.Input((self.max_seq_len, self.dim))
        # Masking layer to handle variable sequence lengths based on zero padding
        masked_inputs = layers.Masking(mask_value=0.0)(inputs)

        h = masked_inputs
        for _ in range(self.num_layers):
            h = layers.GRU(self.hidden_dim, return_sequences=True)(h) if self.module_name == 'gru' else layers.LSTM(self.hidden_dim, return_sequences=True)(h)
        outputs = layers.Dense(self.hidden_dim, activation='sigmoid')(h)
        return keras.Model(inputs=inputs, outputs=outputs, name='embedder')

    def _build_recovery(self):
        inputs = keras.Input((self.max_seq_len, self.hidden_dim))
        h = inputs
        for _ in range(self.num_layers):
            h = layers.GRU(self.hidden_dim, return_sequences=True)(h) if self.module_name == 'gru' else layers.LSTM(self.hidden_dim, return_sequences=True)(h)
        outputs = layers.Dense(self.dim, activation='sigmoid')(h)
        return keras.Model(inputs=inputs, outputs=outputs, name='recovery')

    def _build_generator(self):
        inputs = keras.Input((self.max_seq_len, self.dim)) # z_dim is dim in this context
        h = inputs
        for _ in range(self.num_layers):
            h = layers.GRU(self.hidden_dim, return_sequences=True)(h) if self.module_name == 'gru' else layers.LSTM(self.hidden_dim, return_sequences=True)(h)
        outputs = layers.Dense(self.hidden_dim, activation='sigmoid')(h)
        return keras.Model(inputs=inputs, outputs=outputs, name='generator')

    def _build_supervisor(self):
        inputs = keras.Input((self.max_seq_len, self.hidden_dim))
        h = inputs
        # Original uses num_layers - 1
        for _ in range(self.num_layers - 1):
            h = layers.GRU(self.hidden_dim, return_sequences=True)(h) if self.module_name == 'gru' else layers.LSTM(self.hidden_dim, return_sequences=True)(h)
        outputs = layers.Dense(self.hidden_dim, activation='sigmoid')(h)
        return keras.Model(inputs=inputs, outputs=outputs, name='supervisor')

    def _build_discriminator(self):
        inputs = keras.Input((self.max_seq_len, self.hidden_dim))
        h = inputs
        for _ in range(self.num_layers):
            h = layers.GRU(self.hidden_dim, return_sequences=True)(h) if self.module_name == 'gru' else layers.LSTM(self.hidden_dim, return_sequences=True)(h)
        outputs = layers.Dense(1, activation=None)(h) # No activation for logits
        return keras.Model(inputs=inputs, outputs=outputs, name='discriminator')

    # Loss functions calculations
    def E_loss_T0_fn(self, X_mb, T_mb_list):
        H = self.embedder_model(X_mb)
        X_tilde = self.recovery_model(H)
        loss = self.mse_loss(X_mb, X_tilde)
        return loss

    def E_loss_fn(self, X_mb, T_mb_list):
        # E_loss = E_loss0 + 0.1 * G_loss_S
        H = self.embedder_model(X_mb)
        X_tilde = self.recovery_model(H)
        E_loss_T0 = self.mse_loss(X_mb, X_tilde)
        E_loss0 = 10 * tf.sqrt(E_loss_T0) # Original sqrt was on MSE, which is RMSE

        H_hat_supervise = self.supervisor_model(H)
        # G_loss_S needs H[:,1:,:] and H_hat_supervise[:,:-1,:]
        # Extract based on actual sequence lengths in T_mb_list
        g_loss_s_sum = 0.0
        count = 0
        for i in range(len(T_mb_list)):
            actual_len = T_mb_list[i]
            if actual_len > 1: # G_loss_S requires at least 2 time steps
                g_loss_s_sum += self.mse_loss(H[i, 1:actual_len, :], H_hat_supervise[i, :actual_len-1, :])
                count += 1
        G_loss_S = g_loss_s_sum / count if count > 0 else tf.constant(0.0) # Average loss over batch

        return E_loss0 + 0.1 * G_loss_S

    def D_loss_fn(self, X_mb, Z_mb, T_mb_list):
        H = self.embedder_model(X_mb)
        E_hat = self.generator_model(Z_mb)
        H_hat = self.supervisor_model(E_hat)

        Y_real = self.discriminator_model(H)
        Y_fake = self.discriminator_model(H_hat)
        Y_fake_e = self.discriminator_model(E_hat)

        D_loss_real = self.bce_loss(tf.ones_like(Y_real), Y_real)
        D_loss_fake = self.bce_loss(tf.zeros_like(Y_fake), Y_fake)
        D_loss_fake_e = self.bce_loss(tf.zeros_like(Y_fake_e), Y_fake_e)

        return D_loss_real + D_loss_fake + self.gamma * D_loss_fake_e

    def G_loss_S_fn(self, X_mb, T_mb_list):
        H = self.embedder_model(X_mb)
        H_hat_supervise = self.supervisor_model(H)

        g_loss_s_sum = 0.0
        count = 0
        for i in range(len(T_mb_list)):
            actual_len = T_mb_list[i]
            if actual_len > 1:
                g_loss_s_sum += self.mse_loss(H[i, 1:actual_len, :], H_hat_supervise[i, :actual_len-1, :])
                count += 1
        G_loss_S = g_loss_s_sum / count if count > 0 else tf.constant(0.0)
        return G_loss_S

    def G_loss_fn(self, X_mb, Z_mb, T_mb_list):
        # 1. Adversarial loss
        E_hat = self.generator_model(Z_mb)
        H_hat = self.supervisor_model(E_hat)
        Y_fake = self.discriminator_model(H_hat)
        Y_fake_e = self.discriminator_model(E_hat)

        G_loss_U = self.bce_loss(tf.ones_like(Y_fake), Y_fake)
        G_loss_U_e = self.bce_loss(tf.ones_like(Y_fake_e), Y_fake_e)

        # 2. Supervised loss
        H = self.embedder_model(X_mb)
        H_hat_supervise = self.supervisor_model(H)
        g_loss_s_sum = 0.0
        count = 0
        for i in range(len(T_mb_list)):
            actual_len = T_mb_list[i]
            if actual_len > 1:
                g_loss_s_sum += self.mse_loss(H[i, 1:actual_len, :], H_hat_supervise[i, :actual_len-1, :])
                count += 1
        G_loss_S = g_loss_s_sum / count if count > 0 else tf.constant(0.0)

        # 3. Two Moments
        X_hat = self.recovery_model(H_hat)

        X_unpadded_list = []
        X_hat_unpadded_list = []
        for i in range(len(T_mb_list)):
            actual_len = T_mb_list[i]
            X_unpadded_list.append(X_mb[i, :actual_len, :])
            X_hat_unpadded_list.append(X_hat[i, :actual_len, :])

        if len(X_unpadded_list) > 0:
            X_combined = tf.concat(X_unpadded_list, axis=0)
            X_hat_combined = tf.concat(X_hat_unpadded_list, axis=0)
        else:
            X_combined = tf.zeros((0, self.dim))
            X_hat_combined = tf.zeros((0, self.dim))

        if tf.shape(X_combined)[0] == 0:
            G_loss_V1 = tf.constant(0.0)
            G_loss_V2 = tf.constant(0.0)
        else:
            # Mean and variance across the time-steps and features for all unpadded sequences
            X_mean = tf.reduce_mean(X_combined, axis=0)
            X_var = tf.math.reduce_variance(X_combined, axis=0)
            X_hat_mean = tf.reduce_mean(X_hat_combined, axis=0)
            X_hat_var = tf.math.reduce_variance(X_hat_combined, axis=0)

            G_loss_V1 = tf.reduce_mean(tf.abs(tf.sqrt(X_hat_var + 1e-6) - tf.sqrt(X_var + 1e-6)))
            G_loss_V2 = tf.reduce_mean(tf.abs(X_hat_mean - X_mean))

        G_loss_V = G_loss_V1 + G_loss_V2

        # 4. Summation
        G_loss = G_loss_U + self.gamma * G_loss_U_e + 100 * tf.sqrt(G_loss_S) + 100 * G_loss_V
        return G_loss

    # Training steps using tf.GradientTape
    @tf.function
    def train_step_embedder_initial(self, X_mb_tensor, T_mb_list):
        with tf.GradientTape() as tape:
            e_loss_t0 = self.E_loss_T0_fn(X_mb_tensor, T_mb_list)
            e_loss0 = 10 * tf.sqrt(e_loss_t0)

        e_vars = self.embedder_model.trainable_variables + self.recovery_model.trainable_variables
        gradients = tape.gradient(e_loss0, e_vars)
        self.E0_optimizer.apply_gradients(zip(gradients, e_vars))
        return e_loss_t0

    @tf.function
    def train_step_embedder_joint(self, X_mb_tensor, Z_mb_tensor, T_mb_list):
        with tf.GradientTape() as tape:
            e_loss = self.E_loss_fn(X_mb_tensor, T_mb_list)

        e_vars = self.embedder_model.trainable_variables + self.recovery_model.trainable_variables
        gradients = tape.gradient(e_loss, e_vars)
        self.E_optimizer.apply_gradients(zip(gradients, e_vars))
        return self.E_loss_T0_fn(X_mb_tensor, T_mb_list) # Return E_loss_T0 for printing

    @tf.function
    def train_step_generator_supervised(self, X_mb_tensor, T_mb_list):
        with tf.GradientTape() as tape:
            g_loss_s = self.G_loss_S_fn(X_mb_tensor, T_mb_list)

        g_vars = self.generator_model.trainable_variables + self.supervisor_model.trainable_variables
        gradients = tape.gradient(g_loss_s, g_vars)
        self.GS_optimizer.apply_gradients(zip(gradients, g_vars))
        return g_loss_s

    @tf.function
    def train_step_generator(self, X_mb_tensor, Z_mb_tensor, T_mb_list):
        with tf.GradientTape() as tape:
            g_loss = self.G_loss_fn(X_mb_tensor, Z_mb_tensor, T_mb_list)

        g_vars = self.generator_model.trainable_variables + self.supervisor_model.trainable_variables
        gradients = tape.gradient(g_loss, g_vars)
        self.G_optimizer.apply_gradients(zip(gradients, g_vars))

        # Recalculate components of G_loss for printing outside the tape
        E_hat = self.generator_model(Z_mb_tensor)
        H_hat = self.supervisor_model(E_hat)
        Y_fake = self.discriminator_model(H_hat)
        Y_fake_e = self.discriminator_model(E_hat)
        G_loss_U = self.bce_loss(tf.ones_like(Y_fake), Y_fake)
        G_loss_U_e = self.bce_loss(tf.ones_like(Y_fake_e), Y_fake_e)

        H = self.embedder_model(X_mb_tensor)
        H_hat_supervise = self.supervisor_model(H)
        g_loss_s_sum = 0.0
        count = 0
        for i in range(len(T_mb_list)):
            actual_len = T_mb_list[i]
            if actual_len > 1:
                g_loss_s_sum += self.mse_loss(H[i, 1:actual_len, :], H_hat_supervise[i, :actual_len-1, :])
                count += 1
        G_loss_S = g_loss_s_sum / count if count > 0 else tf.constant(0.0)

        X_hat = self.recovery_model(H_hat)
        X_unpadded_list = []
        X_hat_unpadded_list = []
        for i in range(len(T_mb_list)):
            actual_len = T_mb_list[i]
            X_unpadded_list.append(X_mb_tensor[i, :actual_len, :])
            X_hat_unpadded_list.append(X_hat[i, :actual_len, :])

        if len(X_unpadded_list) > 0:
            X_combined = tf.concat(X_unpadded_list, axis=0)
            X_hat_combined = tf.concat(X_hat_unpadded_list, axis=0)
        else:
            X_combined = tf.zeros((0, self.dim))
            X_hat_combined = tf.zeros((0, self.dim))

        if tf.shape(X_combined)[0] == 0:
            G_loss_V1 = tf.constant(0.0)
            G_loss_V2 = tf.constant(0.0)
        else:
            X_mean = tf.reduce_mean(X_combined, axis=0)
            X_var = tf.math.reduce_variance(X_combined, axis=0)
            X_hat_mean = tf.reduce_mean(X_hat_combined, axis=0)
            X_hat_var = tf.math.reduce_variance(X_hat_combined, axis=0)

            G_loss_V1 = tf.reduce_mean(tf.abs(tf.sqrt(X_hat_var + 1e-6) - tf.sqrt(X_var + 1e-6)))
            G_loss_V2 = tf.reduce_mean(tf.abs(X_hat_mean - X_mean))

        G_loss_V = G_loss_V1 + G_loss_V2

        return G_loss_U, G_loss_S, G_loss_V


    @tf.function
    def train_step_discriminator(self, X_mb_tensor, Z_mb_tensor, T_mb_list):
        with tf.GradientTape() as tape:
            d_loss = self.D_loss_fn(X_mb_tensor, Z_mb_tensor, T_mb_list)

        d_vars = self.discriminator_model.trainable_variables
        gradients = tape.gradient(d_loss, d_vars)
        self.D_optimizer.apply_gradients(zip(gradients, d_vars))
        return d_loss

    @tf.function
    def check_d_loss_val(self, X_mb_tensor, Z_mb_tensor, T_mb_list):
        return self.D_loss_fn(X_mb_tensor, Z_mb_tensor, T_mb_list)

def timegan(ori_data, parameters):
    """TimeGAN function (TensorFlow 2.x Keras version)."""
    # Basic Parameters
    no, seq_len, dim = np.asarray(ori_data).shape

    # Maximum sequence length and each sequence length
    ori_time, max_seq_len = extract_time(ori_data)

    # Normalization
    ori_data_norm, min_val, max_val = MinMaxScaler(ori_data)

    # Pad sequences for consistent input shapes for Keras models
    padded_ori_data = []
    for i in range(len(ori_data_norm)):
        temp = np.zeros([max_seq_len, dim], dtype=np.float32)
        temp[:ori_time[i], :] = ori_data_norm[i]
        padded_ori_data.append(temp)
    padded_ori_data = tf.convert_to_tensor(padded_ori_data)

    # Network Parameters
    hidden_dim   = parameters['hidden_dim']
    num_layers   = parameters['num_layer']
    iterations   = parameters['iterations']
    batch_size   = parameters['batch_size']
    module_name  = parameters['module']
    z_dim        = dim # Latent noise dimension is same as data dimension
    gamma        = 1.0 # Default value

    # Initialize TimeGAN Model
    timegan_model = TimeGANModel(hidden_dim, num_layers, dim, module_name, max_seq_len, gamma)

    # Compile optimizers for the model
    timegan_model.E0_optimizer = keras.optimizers.Adam(learning_rate=1e-3)
    timegan_model.E_optimizer = keras.optimizers.Adam(learning_rate=1e-3)
    timegan_model.G_optimizer = keras.optimizers.Adam(learning_rate=1e-3)
    timegan_model.GS_optimizer = keras.optimizers.Adam(learning_rate=1e-3)
    timegan_model.D_optimizer = keras.optimizers.Adam(learning_rate=1e-3)

    # Call models once to build their layers and create weights
    # This is important for .trainable_variables to be populated
    dummy_input_X = tf.zeros((batch_size, max_seq_len, dim), dtype=tf.float32)
    dummy_input_Z = tf.zeros((batch_size, max_seq_len, z_dim), dtype=tf.float32)
    _ = timegan_model.embedder_model(dummy_input_X)
    _ = timegan_model.recovery_model(timegan_model.embedder_model(dummy_input_X))
    _ = timegan_model.generator_model(dummy_input_Z)
    _ = timegan_model.supervisor_model(timegan_model.generator_model(dummy_input_Z))
    _ = timegan_model.discriminator_model(timegan_model.embedder_model(dummy_input_X))

    # 1. Embedding network training
    print('Start Embedding Network Training')
    for itt in range(iterations):
        X_mb, T_mb = batch_generator(padded_ori_data, ori_time, batch_size)
        e_loss_t0 = timegan_model.train_step_embedder_initial(X_mb, T_mb)
        if itt % 1000 == 0:
            print(f'step: {itt}/{iterations}, e_loss: {np.round(np.sqrt(e_loss_t0.numpy()), 4)}')
    print('Finish Embedding Network Training')

    # 2. Training only with supervised loss
    print('Start Training with Supervised Loss Only')
    for itt in range(iterations):
        X_mb, T_mb = batch_generator(padded_ori_data, ori_time, batch_size)
        g_loss_s = timegan_model.train_step_generator_supervised(X_mb, T_mb)
        if itt % 1000 == 0:
            print(f'step: {itt}/{iterations}, s_loss: {np.round(np.sqrt(g_loss_s.numpy()), 4)}')
    print('Finish Training with Supervised Loss Only')

    # 3. Joint Training
    print('Start Joint Training')
    for itt in range(iterations):
        # Generator training (twice more than discriminator training)
        for _ in range(2):
            X_mb, T_mb = batch_generator(padded_ori_data, ori_time, batch_size)
            Z_mb = random_generator(batch_size, z_dim, T_mb, max_seq_len)

            g_loss_u, g_loss_s, g_loss_v = timegan_model.train_step_generator(X_mb, Z_mb, T_mb)
            e_loss_t0 = timegan_model.train_step_embedder_joint(X_mb, Z_mb, T_mb)

        # Discriminator training
        X_mb, T_mb = batch_generator(padded_ori_data, ori_time, batch_size)
        Z_mb = random_generator(batch_size, z_dim, T_mb, max_seq_len)

        check_d_loss = timegan_model.check_d_loss_val(X_mb, Z_mb, T_mb)
        # Ensure d_loss is always a tensor for consistent printing
        d_loss = check_d_loss # Initialize with check_d_loss

        if check_d_loss.numpy() > 0.15:
            d_loss = timegan_model.train_step_discriminator(X_mb, Z_mb, T_mb)

        if itt % 1000 == 0:
            print(f'step: {itt}/{iterations}, d_loss: {np.round(d_loss.numpy(), 4)}, '
                  f'g_loss_u: {np.round(g_loss_u.numpy(), 4)}, '
                  f'g_loss_s: {np.round(np.sqrt(g_loss_s.numpy()), 4)}, '
                  f'g_loss_v: {np.round(g_loss_v.numpy(), 4)}, '
                  f'e_loss_t0: {np.round(np.sqrt(e_loss_t0.numpy()), 4)}')
    print('Finish Joint Training')

    ## Synthetic data generation
    # Generate Z for all data points
    Z_full = random_generator(no, z_dim, ori_time, max_seq_len)

    # Forward pass through Generator and Supervisor
    E_hat_full = timegan_model.generator_model(Z_full)
    H_hat_full = timegan_model.supervisor_model(E_hat_full)

    # Recover synthetic data
    X_hat_full = timegan_model.recovery_model(H_hat_full)

    generated_data_curr = X_hat_full.numpy()

    generated_data = list()
    for i in range(no):
        temp = generated_data_curr[i, :ori_time[i], :]
        generated_data.append(temp)

    # Renormalization
    generated_data = np.asarray(generated_data) * max_val
    generated_data = generated_data + min_val

    return generated_data

with open('timegan_tf2.py', 'w') as f:
    f.write('''
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np

# Helper functions - adapted for TF2/Keras
def extract_time(data):
    """Returns max_seq_len and a list of sequence lengths for each sample."""
    time = list()
    max_seq_len = 0
    for i in range(len(data)):
        max_seq_len = max(max_seq_len, len(data[i]))
        time.append(len(data[i]))
    return time, max_seq_len

def random_generator(batch_size, z_dim, T_mb, max_seq_len):
    """Random vector generation.

    Args:
      - batch_size: size of the random vector
      - z_dim: dimension of the random vector
      - T_mb: time information (list of actual sequence lengths)
      - max_seq_len: maximum sequence length (for padding)

    Returns:
      - Z_mb: generated random vector (padded)
    """
    Z_mb = []
    for i in range(batch_size):
        temp = np.zeros([max_seq_len, z_dim], dtype=np.float32)
        temp_Z = np.random.uniform(0., 1, [T_mb[i], z_dim]).astype(np.float32)
        temp[:T_mb[i],:] = temp_Z
        Z_mb.append(temp)
    return tf.convert_to_tensor(Z_mb)

def batch_generator(data, time, batch_size):
    """Mini-batch generator.

    Args:
      - data: original time-series data (normalized and padded)
      - time: list of original sequence lengths
      - batch_size: mini-batch size

    Returns:
      - X_mb: time-series data in mini-batch (padded tensor)
      - T_mb: time information in mini-batch (list of original sequence lengths)
    """
    no = len(data)
    idx = np.random.permutation(no)
    train_idx = idx[:batch_size]

    X_mb_list = []
    T_mb_list = []
    for i in range(batch_size):
        X_mb_list.append(data[train_idx[i]])
        T_mb_list.append(time[train_idx[i]])

    return tf.convert_to_tensor(X_mb_list), T_mb_list

def MinMaxScaler(data):
    """Min-Max Normalizer.

    Args:
      - data: raw data

    Returns:
      - norm_data: normalized data
      - min_val: minimum values (for renormalization)
      - max_val: maximum values (for renormalization)
    """
    min_val = np.min(np.min(data, axis = 0), axis = 0)
    data = data - min_val

    max_val = np.max(np.max(data, axis = 0), axis = 0)
    norm_data = data / (max_val + 1e-7)

    return norm_data, min_val, max_val

class TimeGANModel(keras.Model):
    def __init__(self, hidden_dim, num_layers, dim, module_name='gru', max_seq_len=None, gamma=1.0):
        super(TimeGANModel, self).__init__()
        self.hidden_dim = hidden_dim
        self.num_layers = num_layers
        self.dim = dim
        self.module_name = module_name
        self.max_seq_len = max_seq_len
        self.gamma = gamma

        # Models
        self.embedder_model = self._build_embedder()
        self.recovery_model = self._build_recovery()
        self.generator_model = self._build_generator()
        self.supervisor_model = self._build_supervisor()
        self.discriminator_model = self._build_discriminator()

        # Optimizers (will be instantiated later in the training loop)
        self.E0_optimizer = None
        self.E_optimizer = None
        self.G_optimizer = None
        self.GS_optimizer = None
        self.D_optimizer = None

        # Losses
        self.mse_loss = keras.losses.MeanSquaredError()
        self.bce_loss = keras.losses.BinaryCrossentropy(from_logits=True)

    def _build_embedder(self):
        inputs = keras.Input((self.max_seq_len, self.dim))
        masked_inputs = layers.Masking(mask_value=0.0)(inputs)

        h = masked_inputs
        for _ in range(self.num_layers):
            h = layers.GRU(self.hidden_dim, return_sequences=True)(h) if self.module_name == 'gru' else layers.LSTM(self.hidden_dim, return_sequences=True)(h)
        outputs = layers.Dense(self.hidden_dim, activation='sigmoid')(h)
        return keras.Model(inputs=inputs, outputs=outputs, name='embedder')

    def _build_recovery(self):
        inputs = keras.Input((self.max_seq_len, self.hidden_dim))
        h = inputs
        for _ in range(self.num_layers):
            h = layers.GRU(self.hidden_dim, return_sequences=True)(h) if self.module_name == 'gru' else layers.LSTM(self.hidden_dim, return_sequences=True)(h)
        outputs = layers.Dense(self.dim, activation='sigmoid')(h)
        return keras.Model(inputs=inputs, outputs=outputs, name='recovery')

    def _build_generator(self):
        inputs = keras.Input((self.max_seq_len, self.dim))
        h = inputs
        for _ in range(self.num_layers):
            h = layers.GRU(self.hidden_dim, return_sequences=True)(h) if self.module_name == 'gru' else layers.LSTM(self.hidden_dim, return_sequences=True)(h)
        outputs = layers.Dense(self.hidden_dim, activation='sigmoid')(h)
        return keras.Model(inputs=inputs, outputs=outputs, name='generator')

    def _build_supervisor(self):
        inputs = keras.Input((self.max_seq_len, self.hidden_dim))
        h = inputs
        for _ in range(self.num_layers - 1):
            h = layers.GRU(self.hidden_dim, return_sequences=True)(h) if self.module_name == 'gru' else layers.LSTM(self.hidden_dim, return_sequences=True)(h)
        outputs = layers.Dense(self.hidden_dim, activation='sigmoid')(h)
        return keras.Model(inputs=inputs, outputs=outputs, name='supervisor')

    def _build_discriminator(self):
        inputs = keras.Input((self.max_seq_len, self.hidden_dim))
        h = inputs
        for _ in range(self.num_layers):
            h = layers.GRU(self.hidden_dim, return_sequences=True)(h) if self.module_name == 'gru' else layers.LSTM(self.hidden_dim, return_sequences=True)(h)
        outputs = layers.Dense(1, activation=None)(h)
        return keras.Model(inputs=inputs, outputs=outputs, name='discriminator')

    def E_loss_T0_fn(self, X_mb, T_mb_list):
        H = self.embedder_model(X_mb)
        X_tilde = self.recovery_model(H)
        loss = self.mse_loss(X_mb, X_tilde)
        return loss

    def E_loss_fn(self, X_mb, T_mb_list):
        H = self.embedder_model(X_mb)
        X_tilde = self.recovery_model(H)
        E_loss_T0 = self.mse_loss(X_mb, X_tilde)
        E_loss0 = 10 * tf.sqrt(E_loss_T0)

        H_hat_supervise = self.supervisor_model(H)
        g_loss_s_sum = 0.0
        count = 0
        for i in range(len(T_mb_list)):
            actual_len = T_mb_list[i]
            if actual_len > 1:
                g_loss_s_sum += self.mse_loss(H[i, 1:actual_len, :], H_hat_supervise[i, :actual_len-1, :])
                count += 1
        G_loss_S = g_loss_s_sum / count if count > 0 else tf.constant(0.0)

        return E_loss0 + 0.1 * G_loss_S

    def D_loss_fn(self, X_mb, Z_mb, T_mb_list):
        H = self.embedder_model(X_mb)
        E_hat = self.generator_model(Z_mb)
        H_hat = self.supervisor_model(E_hat)

        Y_real = self.discriminator_model(H)
        Y_fake = self.discriminator_model(H_hat)
        Y_fake_e = self.discriminator_model(E_hat)

        D_loss_real = self.bce_loss(tf.ones_like(Y_real), Y_real)
        D_loss_fake = self.bce_loss(tf.zeros_like(Y_fake), Y_fake)
        D_loss_fake_e = self.bce_loss(tf.zeros_like(Y_fake_e), Y_fake_e)

        return D_loss_real + D_loss_fake + self.gamma * D_loss_fake_e

    def G_loss_S_fn(self, X_mb, T_mb_list):
        H = self.embedder_model(X_mb)
        H_hat_supervise = self.supervisor_model(H)

        g_loss_s_sum = 0.0
        count = 0
        for i in range(len(T_mb_list)):
            actual_len = T_mb_list[i]
            if actual_len > 1:
                g_loss_s_sum += self.mse_loss(H[i, 1:actual_len, :], H_hat_supervise[i, :actual_len-1, :])
                count += 1
        G_loss_S = g_loss_s_sum / count if count > 0 else tf.constant(0.0)
        return G_loss_S

    def G_loss_fn(self, X_mb, Z_mb, T_mb_list):
        # 1. Adversarial loss
        E_hat = self.generator_model(Z_mb)
        H_hat = self.supervisor_model(E_hat)
        Y_fake = self.discriminator_model(H_hat)
        Y_fake_e = self.discriminator_model(E_hat)

        G_loss_U = self.bce_loss(tf.ones_like(Y_fake), Y_fake)
        G_loss_U_e = self.bce_loss(tf.ones_like(Y_fake_e), Y_fake_e)

        # 2. Supervised loss
        H = self.embedder_model(X_mb)
        H_hat_supervise = self.supervisor_model(H)
        g_loss_s_sum = 0.0
        count = 0
        for i in range(len(T_mb_list)):
            actual_len = T_mb_list[i]
            if actual_len > 1:
                g_loss_s_sum += self.mse_loss(H[i, 1:actual_len, :], H_hat_supervise[i, :actual_len-1, :])
                count += 1
        G_loss_S = g_loss_s_sum / count if count > 0 else tf.constant(0.0)

        # 3. Two Moments
        X_hat = self.recovery_model(H_hat)

        X_unpadded_list = []
        X_hat_unpadded_list = []
        for i in range(len(T_mb_list)):
            actual_len = T_mb_list[i]
            X_unpadded_list.append(X_mb[i, :actual_len, :])
            X_hat_unpadded_list.append(X_hat[i, :actual_len, :])

        if len(X_unpadded_list) > 0:
            X_combined = tf.concat(X_unpadded_list, axis=0)
            X_hat_combined = tf.concat(X_hat_unpadded_list, axis=0)
        else:
            X_combined = tf.zeros((0, self.dim))
            X_hat_combined = tf.zeros((0, self.dim))

        if tf.shape(X_combined)[0] == 0:
            G_loss_V1 = tf.constant(0.0)
            G_loss_V2 = tf.constant(0.0)
        else:
            X_mean = tf.reduce_mean(X_combined, axis=0)
            X_var = tf.math.reduce_variance(X_combined, axis=0)
            X_hat_mean = tf.reduce_mean(X_hat_combined, axis=0)
            X_hat_var = tf.math.reduce_variance(X_hat_combined, axis=0)

            G_loss_V1 = tf.reduce_mean(tf.abs(tf.sqrt(X_hat_var + 1e-6) - tf.sqrt(X_var + 1e-6)))
            G_loss_V2 = tf.reduce_mean(tf.abs(X_hat_mean - X_mean))

        G_loss_V = G_loss_V1 + G_loss_V2

        # 4. Summation
        G_loss = G_loss_U + self.gamma * G_loss_U_e + 100 * tf.sqrt(G_loss_S) + 100 * G_loss_V
        return G_loss

    @tf.function
    def train_step_embedder_initial(self, X_mb_tensor, T_mb_list):
        with tf.GradientTape() as tape:
            e_loss_t0 = self.E_loss_T0_fn(X_mb_tensor, T_mb_list)
            e_loss0 = 10 * tf.sqrt(e_loss_t0)

        e_vars = self.embedder_model.trainable_variables + self.recovery_model.trainable_variables
        gradients = tape.gradient(e_loss0, e_vars)
        self.E0_optimizer.apply_gradients(zip(gradients, e_vars))
        return e_loss_t0

    @tf.function
    def train_step_embedder_joint(self, X_mb_tensor, Z_mb_tensor, T_mb_list):
        with tf.GradientTape() as tape:
            e_loss = self.E_loss_fn(X_mb_tensor, T_mb_list)

        e_vars = self.embedder_model.trainable_variables + self.recovery_model.trainable_variables
        gradients = tape.gradient(e_loss, e_vars)
        self.E_optimizer.apply_gradients(zip(gradients, e_vars))
        return self.E_loss_T0_fn(X_mb_tensor, T_mb_list)

    @tf.function
    def train_step_generator_supervised(self, X_mb_tensor, T_mb_list):
        with tf.GradientTape() as tape:
            g_loss_s = self.G_loss_S_fn(X_mb_tensor, T_mb_list)

        g_vars = self.generator_model.trainable_variables + self.supervisor_model.trainable_variables
        gradients = tape.gradient(g_loss_s, g_vars)
        self.GS_optimizer.apply_gradients(zip(gradients, g_vars))
        return g_loss_s

    @tf.function
    def train_step_generator(self, X_mb_tensor, Z_mb_tensor, T_mb_list):
        with tf.GradientTape() as tape:
            g_loss = self.G_loss_fn(X_mb_tensor, Z_mb_tensor, T_mb_list)

        g_vars = self.generator_model.trainable_variables + self.supervisor_model.trainable_variables
        gradients = tape.gradient(g_loss, g_vars)
        self.G_optimizer.apply_gradients(zip(gradients, g_vars))

        E_hat = self.generator_model(Z_mb_tensor)
        H_hat = self.supervisor_model(E_hat)
        Y_fake = self.discriminator_model(H_hat)
        Y_fake_e = self.discriminator_model(E_hat)
        G_loss_U = self.bce_loss(tf.ones_like(Y_fake), Y_fake)
        G_loss_U_e = self.bce_loss(tf.ones_like(Y_fake_e), Y_fake_e)

        H = self.embedder_model(X_mb_tensor)
        H_hat_supervise = self.supervisor_model(H)
        g_loss_s_sum = 0.0
        count = 0
        for i in range(len(T_mb_list)):
            actual_len = T_mb_list[i]
            if actual_len > 1:
                g_loss_s_sum += self.mse_loss(H[i, 1:actual_len, :], H_hat_supervise[i, :actual_len-1, :])
                count += 1
        G_loss_S = g_loss_s_sum / count if count > 0 else tf.constant(0.0)

        X_hat = self.recovery_model(H_hat)
        X_unpadded_list = []
        X_hat_unpadded_list = []
        for i in range(len(T_mb_list)):
            actual_len = T_mb_list[i]
            X_unpadded_list.append(X_mb_tensor[i, :actual_len, :])
            X_hat_unpadded_list.append(X_hat[i, :actual_len, :])

        if len(X_unpadded_list) > 0:
            X_combined = tf.concat(X_unpadded_list, axis=0)
            X_hat_combined = tf.concat(X_hat_unpadded_list, axis=0)
        else:
            X_combined = tf.zeros((0, self.dim))
            X_hat_combined = tf.zeros((0, self.dim))

        if tf.shape(X_combined)[0] == 0:
            G_loss_V1 = tf.constant(0.0)
            G_loss_V2 = tf.constant(0.0)
        else:
            X_mean = tf.reduce_mean(X_combined, axis=0)
            X_var = tf.math.reduce_variance(X_combined, axis=0)
            X_hat_mean = tf.reduce_mean(X_hat_combined, axis=0)
            X_hat_var = tf.math.reduce_variance(X_hat_combined, axis=0)

            G_loss_V1 = tf.reduce_mean(tf.abs(tf.sqrt(X_hat_var + 1e-6) - tf.sqrt(X_var + 1e-6)))
            G_loss_V2 = tf.reduce_mean(tf.abs(X_hat_mean - X_mean))

        G_loss_V = G_loss_V1 + G_loss_V2

        return G_loss_U, G_loss_S, G_loss_V


    @tf.function
    def train_step_discriminator(self, X_mb_tensor, Z_mb_tensor, T_mb_list):
        with tf.GradientTape() as tape:
            d_loss = self.D_loss_fn(X_mb_tensor, Z_mb_tensor, T_mb_list)

        d_vars = self.discriminator_model.trainable_variables
        gradients = tape.gradient(d_loss, d_vars)
        self.D_optimizer.apply_gradients(zip(gradients, d_vars))
        return d_loss

    @tf.function
    def check_d_loss_val(self, X_mb_tensor, Z_mb_tensor, T_mb_list):
        return self.D_loss_fn(X_mb_tensor, Z_mb_tensor, T_mb_list)

def timegan(ori_data, parameters):
    """TimeGAN function (TensorFlow 2.x Keras version)."""
    # Basic Parameters
    no, seq_len, dim = np.asarray(ori_data).shape

    # Maximum sequence length and each sequence length
    ori_time, max_seq_len = extract_time(ori_data)

    # Normalization
    ori_data_norm, min_val, max_val = MinMaxScaler(ori_data)

    # Pad sequences for consistent input shapes for Keras models
    padded_ori_data = []
    for i in range(len(ori_data_norm)):
        temp = np.zeros([max_seq_len, dim], dtype=np.float32)
        temp[:ori_time[i], :] = ori_data_norm[i]
        padded_ori_data.append(temp)
    padded_ori_data = tf.convert_to_tensor(padded_ori_data)

    # Network Parameters
    hidden_dim   = parameters['hidden_dim']
    num_layers   = parameters['num_layer']
    iterations   = parameters['iterations']
    batch_size   = parameters['batch_size']
    module_name  = parameters['module']
    z_dim        = dim # Latent noise dimension is same as data dimension
    gamma        = 1.0 # Default value

    # Initialize TimeGAN Model
    timegan_model = TimeGANModel(hidden_dim, num_layers, dim, module_name, max_seq_len, gamma)

    # Compile optimizers for the model
    timegan_model.E0_optimizer = keras.optimizers.Adam(learning_rate=1e-3)
    timegan_model.E_optimizer = keras.optimizers.Adam(learning_rate=1e-3)
    timegan_model.G_optimizer = keras.optimizers.Adam(learning_rate=1e-3)
    timegan_model.GS_optimizer = keras.optimizers.Adam(learning_rate=1e-3)
    timegan_model.D_optimizer = keras.optimizers.Adam(learning_rate=1e-3)

    # Call models once to build their layers and create weights
    dummy_input_X = tf.zeros((batch_size, max_seq_len, dim), dtype=tf.float32)
    dummy_input_Z = tf.zeros((batch_size, max_seq_len, z_dim), dtype=tf.float32)
    _ = timegan_model.embedder_model(dummy_input_X)
    _ = timegan_model.recovery_model(timegan_model.embedder_model(dummy_input_X))
    _ = timegan_model.generator_model(dummy_input_Z)
    _ = timegan_model.supervisor_model(timegan_model.generator_model(dummy_input_Z))
    _ = timegan_model.discriminator_model(timegan_model.embedder_model(dummy_input_X))

    # 1. Embedding network training
    print('Start Embedding Network Training')
    for itt in range(iterations):
        X_mb, T_mb = batch_generator(padded_ori_data, ori_time, batch_size)
        e_loss_t0 = timegan_model.train_step_embedder_initial(X_mb, T_mb)
        if itt % 1000 == 0:
            print(f'step: {itt}/{iterations}, e_loss: {np.round(np.sqrt(e_loss_t0.numpy()), 4)}')
    print('Finish Embedding Network Training')

    # 2. Training only with supervised loss
    print('Start Training with Supervised Loss Only')
    for itt in range(iterations):
        X_mb, T_mb = batch_generator(padded_ori_data, ori_time, batch_size)
        g_loss_s = timegan_model.train_step_generator_supervised(X_mb, T_mb)
        if itt % 1000 == 0:
            print(f'step: {itt}/{iterations}, s_loss: {np.round(np.sqrt(g_loss_s.numpy()), 4)}')
    print('Finish Training with Supervised Loss Only')

    # 3. Joint Training
    print('Start Joint Training')
    for itt in range(iterations):
        # Generator training (twice more than discriminator training)
        for _ in range(2):
            X_mb, T_mb = batch_generator(padded_ori_data, ori_time, batch_size)
            Z_mb = random_generator(batch_size, z_dim, T_mb, max_seq_len)

            g_loss_u, g_loss_s, g_loss_v = timegan_model.train_step_generator(X_mb, Z_mb, T_mb)
            e_loss_t0 = timegan_model.train_step_embedder_joint(X_mb, Z_mb, T_mb)

        # Discriminator training
        X_mb, T_mb = batch_generator(padded_ori_data, ori_time, batch_size)
        Z_mb = random_generator(batch_size, z_dim, T_mb, max_seq_len)

        check_d_loss = timegan_model.check_d_loss_val(X_mb, Z_mb, T_mb)
        d_loss = check_d_loss

        if check_d_loss.numpy() > 0.15:
            d_loss = timegan_model.train_step_discriminator(X_mb, Z_mb, T_mb)

        if itt % 1000 == 0:
            print(f'step: {itt}/{iterations}, d_loss: {np.round(d_loss.numpy(), 4)}, '
                  f'g_loss_u: {np.round(g_loss_u.numpy(), 4)}, '
                  f'g_loss_s: {np.round(np.sqrt(g_loss_s.numpy()), 4)}, '
                  f'g_loss_v: {np.round(g_loss_v.numpy(), 4)}, '
                  f'e_loss_t0: {np.round(np.sqrt(e_loss_t0.numpy()), 4)}')
    print('Finish Joint Training')

    ## Synthetic data generation
    Z_full = random_generator(no, z_dim, ori_time, max_seq_len)

    E_hat_full = timegan_model.generator_model(Z_full)
    H_hat_full = timegan_model.supervisor_model(E_hat_full)

    X_hat_full = timegan_model.recovery_model(H_hat_full)

    generated_data_curr = X_hat_full.numpy()

    generated_data = list()
    for i in range(no):
        temp = generated_data_curr[i, :ori_time[i], :]
        generated_data.append(temp)

    # Renormalization
    generated_data = np.asarray(generated_data) * max_val
    generated_data = generated_data + min_val

    return generated_data
''')
print("Content written to timegan_tf2.py")
```
