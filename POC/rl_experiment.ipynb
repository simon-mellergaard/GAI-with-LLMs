# RL in energy trading POC

Using RL to take trading decisions based on the current state.


Based on the data, make a reinforcement learning agent that makes trading decisions. The RL setup is as follows:
Actions:

* Buy
* Sell
* Do nothing

If the agent choses buy while in the 'do nothing' state, the agent can only chose buy again after returning to the 'do nothing' state. If the agent choses sell while in the 'do nothing' state, the agent can only sell again after returning to the 'do nothing' state.
State:

* position: (1,0,-1)
* pos_price: continuous
* cur_price: continuous
* last_price: continuous

The state is 1 if the agent choses buy while in the 0 state, and -1 if the agent choses sell in the 0 state. The state is 0 if the agent choses sell in the sell in the 1 state, and if the agent choses buy in the -1 state.
The cur_price is the current price at the timestamp using the data dataframe, and the last_price is the previous price. pos_price is the price at the moment the agent choses buy or sell while in state 0.

Rewards:

* When in state 1 and chosing action sell: cur_price - pos_price
* When in state -1 and chosing action buy: pos_price - cur_price
* All other states: 0

Make a simple implementation of this agent and environment using the data in the dataframe 'data'.

## Define the Trading Environment Class

### Subtask:
Create a Python class named `TradingEnvironment` that simulates the trading environment. This class will manage the state, process actions, calculate rewards, and handle transitions using the provided `data` DataFrame.


**Reasoning**:
First, I will define the `TradingEnvironment` class with its `__init__` method. This method will initialize the environment by taking the `data` DataFrame, storing the real prices, setting the initial step, and determining the maximum number of steps. It will also define the possible actions.



```python
import numpy as np

class TradingEnvironment:
    def __init__(self, data):
        self.data = data
        self.prices = self.data['Real price'].values
        self.max_steps = len(self.prices) - 1  # -1 because we need last_price

        self.action_space = ['buy', 'sell', 'do_nothing']

        # Initialize environment state
        self.current_step = 0
        self.position = 0  # 0: do nothing, 1: buy, -1: sell
        self.pos_price = 0  # price at which position was opened
        self.cur_price = 0
        self.last_price = 0

        print("TradingEnvironment initialized.")

    def reset(self):
        self.current_step = 0
        self.position = 0
        self.pos_price = 0
        self.cur_price = self.prices[self.current_step]
        self.last_price = self.prices[self.current_step] # For the very first step, last_price can be same as current_price

        # State: (position, pos_price, cur_price, last_price)
        return (self.position, self.pos_price, self.cur_price, self.last_price)

    def step(self, action):
        self.current_step += 1
        if self.current_step > self.max_steps:
            # If episode is done, return the last state with 0 reward
            return (self.position, self.pos_price, self.cur_price, self.last_price), 0, True

        self.last_price = self.cur_price
        self.cur_price = self.prices[self.current_step]

        reward = 0
        done = False

        if self.current_step == self.max_steps:
            done = True

        # Apply trading rules based on current position and action
        if action == 'buy':
            if self.position == 0: # Can only buy if not holding any position
                self.position = 1
                self.pos_price = self.cur_price
            elif self.position == -1: # Close short position
                reward = self.pos_price - self.cur_price # Buy to cover short
                self.position = 0
                self.pos_price = 0
            # If self.position == 1 (already bought), do nothing implicitly based on problem description

        elif action == 'sell':
            if self.position == 0: # Can only sell if not holding any position
                self.position = -1
                self.pos_price = self.cur_price
            elif self.position == 1: # Close long position
                reward = self.cur_price - self.pos_price # Sell to close long
                self.position = 0
                self.pos_price = 0
            # If self.position == -1 (already sold), do nothing implicitly based on problem description

        # 'do_nothing' action always results in 0 reward and no change in position/pos_price unless closing a position

        return (self.position, self.pos_price, self.cur_price, self.last_price), reward, done
```

## Implement a Simple RL Agent

### Subtask:
Implement a basic reinforcement learning agent that can interact with the defined environment. This agent will decide on actions based on the current state and learn from the rewards.


**Reasoning**:
I will define the `RLAgent` class with its constructor, `_discretize_state`, `act`, `learn`, and `decay_epsilon` methods according to the instructions, implementing the core logic for the reinforcement learning agent.



```python
import random

class RLAgent:
    def __init__(self, action_space, alpha=0.1, gamma=0.9, epsilon=1.0, epsilon_decay_rate=0.995):
        self.action_space = action_space
        self.alpha = alpha  # Learning rate
        self.gamma = gamma  # Discount factor
        self.epsilon = epsilon  # Exploration-exploitation trade-off
        self.epsilon_decay_rate = epsilon_decay_rate # Epsilon decay rate
        self.q_table = {}  # Q-table to store Q-values

        print("RLAgent initialized.")

    def _discretize_state(self, state):
        position, pos_price, cur_price, last_price = state

        # Discretize price trend
        if cur_price > last_price:
            price_trend = 'up'
        elif cur_price < last_price:
            price_trend = 'down'
        else:
            price_trend = 'flat'

        # Discretize profit/loss status if holding a position
        profit_loss_status = 'no_position'
        if position == 1: # Long position
            if cur_price > pos_price:
                profit_loss_status = 'profit'
            elif cur_price < pos_price:
                profit_loss_status = 'loss'
            else:
                profit_loss_status = 'break_even'
        elif position == -1: # Short position
            if cur_price < pos_price:
                profit_loss_status = 'profit'
            elif cur_price > pos_price:
                profit_loss_status = 'loss'
            else:
                profit_loss_status = 'break_even'

        # Combine discrete features into a hashable tuple
        discretized_state = (position, price_trend, profit_loss_status)
        return discretized_state

    def act(self, state):
        discretized_state = self._discretize_state(state)

        if random.uniform(0, 1) < self.epsilon:
            # Explore: choose a random action
            return random.choice(self.action_space)
        else:
            # Exploit: choose the action with the highest Q-value
            if discretized_state not in self.q_table:
                # If state is new, initialize Q-values to 0
                self.q_table[discretized_state] = {action: 0.0 for action in self.action_space}

            q_values = self.q_table[discretized_state]
            # Choose action with max Q-value, break ties randomly
            max_q = max(q_values.values())
            actions_with_max_q = [action for action, q_val in q_values.items() if q_val == max_q]
            return random.choice(actions_with_max_q)

    def learn(self, state, action, reward, next_state, done):
        discretized_state = self._discretize_state(state)
        discretized_next_state = self._discretize_state(next_state)

        # Get current Q-value for (state, action)
        if discretized_state not in self.q_table:
            self.q_table[discretized_state] = {a: 0.0 for a in self.action_space}
        current_q = self.q_table[discretized_state][action]

        # Calculate max Q-value for next_state
        if done:
            max_next_q = 0 # No future reward if episode is done
        else:
            if discretized_next_state not in self.q_table:
                self.q_table[discretized_next_state] = {a: 0.0 for a in self.action_space}
            max_next_q = max(self.q_table[discretized_next_state].values())

        # Q-learning update rule
        new_q = current_q + self.alpha * (reward + self.gamma * max_next_q - current_q)
        self.q_table[discretized_state][action] = new_q

    def decay_epsilon(self):
        self.epsilon *= self.epsilon_decay_rate
```

## Simulate and Evaluate Trading

### Subtask:
Run a simulation where the implemented agent interacts with the environment over the entire 'data' dataframe. Track the agent's actions, state transitions, and accumulated rewards. Evaluate the agent's performance by calculating the total profit or loss over the simulation period.


**Reasoning**:
I will implement the simulation loop to train the `RLAgent` within the `TradingEnvironment`, track rewards, and then evaluate its performance by calculating total and average profit/loss.



```python
env = TradingEnvironment(data)
agent = RLAgent(env.action_space)

num_episodes = 200 # Increased for better training
total_rewards_per_episode = []

print(f"Starting simulation for {num_episodes} episodes...")

for episode in range(num_episodes):
    state = env.reset()
    episode_reward = 0
    done = False

    while not done:
        action = agent.act(state)
        next_state, reward, done = env.step(action)
        agent.learn(state, action, reward, next_state, done)
        episode_reward += reward
        state = next_state

    agent.decay_epsilon()
    total_rewards_per_episode.append(episode_reward)

    if (episode + 1) % 50 == 0:
        print(f"Episode {episode + 1}/{num_episodes} completed. Total reward: {episode_reward:.2f}")

# Evaluate agent's performance
average_profit = np.mean(total_rewards_per_episode)
last_episode_profit = total_rewards_per_episode[-1]

print("\n--- Simulation Results ---")
print(f"Total episodes: {num_episodes}")
print(f"Average profit/loss per episode: {average_profit:.2f}")
print(f"Total profit/loss in the last episode: {last_episode_profit:.2f}")
print(f"Final Q-table size: {len(agent.q_table)}")
```

## Detailed Trading Decisions for a Single Episode

To observe the agent's trading decisions, let's run a single episode and print the state, chosen action, and received reward at each step.

```python
env_single_run = TradingEnvironment(data)
# Re-initialize agent for a fresh run, or use the trained agent's Q-table if desired
# For demonstration, let's use the trained agent but with exploration turned off (epsilon=0)
# to see its learned policy.
# Or, if you want to see exploration in action, use a new agent with default epsilon
agent_single_run = RLAgent(env_single_run.action_space, epsilon=0.0) # Use the learned policy

# Copy the learned Q-table from the trained agent
agent_single_run.q_table = agent.q_table.copy()

print("--- Running a single episode with learned policy ---")

state = env_single_run.reset()
episode_reward = 0
done = False
step_count = 0

trading_log = [] # List to store trading decisions

while not done:
    action = agent_single_run.act(state)
    next_state, reward, done = env_single_run.step(action)
    episode_reward += reward

    position, pos_price, cur_price, last_price = state
    next_position, next_pos_price, next_cur_price, next_last_price = next_state

    # Log the details for the current step
    trading_log.append({
        'step': step_count,
        'position': position,
        'pos_price': pos_price,
        'cur_price': cur_price,
        'last_price': last_price,
        'action_taken': action,
        'reward': reward,
        'next_position': next_position,
        'next_pos_price': next_pos_price,
        'next_cur_price': next_cur_price,
        'next_last_price': next_last_price
    })

    state = next_state
    step_count += 1

print(f"Single episode finished. Total reward: {episode_reward:.2f}")

# Convert the log to a pandas DataFrame
trading_df = pd.DataFrame(trading_log)
print("\n--- Trading Decisions DataFrame ---")
display(trading_df.head())
```
